`Reshaping and stacking tensors` \
1. `Reshaping` - reshapes input tensor to defined shape
2. `View` - Return a view of input tensor of certain shape but keep same memory as parent tensor
3. `Stack` - combine tensors on top of each other (vstack) or side by side (hstack)
4. `Squeeze` - remove all '1' dimensions from a tensor
5. `Unsqueeze` - adds a '1' dimensions to target tensor
6. `Permute` - Return a view of input tensor with dimensions swapped in a certain way

```python
import torch as t
x=t.arange(0,10)
x, x.shape
```

```python
#reshape
#x_reshaped = x.reshape(1,10)
#x_reshaped = x.reshape(10,1)
x_reshaped = x.reshape(5,2)
x_reshaped, x_reshaped.shape
```

`x` and `z` share the same memory location therefore any change made to `z` will reflect in `x`

```python
#change the view
z= x.view(10,1)
z, z.shape
```

```python
#stacking
x_stacked = t.stack((x,x,x,x), dim=0) #vertical stacking
print(x_stacked, x_stacked.shape)
x_stacked_horizontal = t.stack((x,x,x,x), dim=1)
print(x_stacked_horizontal, x_stacked_horizontal.shape)
```

```python
#squeezing and unsqueezing
x_reshaped_new=x.reshape(1,10)
x_reshaped_new,x_reshaped_new.shape
```

```python
x_reshaped_new.squeeze(), x_reshaped_new.squeeze().shape
```

```python
x_reshaped_new.unsqueeze(dim=1), x_reshaped_new.unsqueeze(dim=1).shape
```

`permuting`

```python
y=t.rand(2,3,3)
print(y, y.shape)
y_p=t.permute(y,(2,1,0))
print(y_p, y_p.shape)
```
