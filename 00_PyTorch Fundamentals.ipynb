**Zeros and Ones Tensors**

```python
X=torch.zeros(2,2,3)
Y=torch.ones(2,2,3)
print(X), print(Y)
```

```python
Y_new=torch.rand_like(Y) #generate a tensor that is dimensionally same as Y, but with random numbers
print(Y_new)
```
