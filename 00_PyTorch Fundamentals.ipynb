```python
rangedTensor=torch.arange(start=1, end=1000, step=20.5)
rangedTensor
```

```python
rangedZeroTensor=torch.rand_like(rangedTensor) #random tensor same size as rangedTensor
rangedZeroTensor
```

### Tensor Datatypes

```python
rand_tensor= torch.rand_like(rangedZeroTensor, dtype=None, device=None,requires_grad=False)
#dtype is float32 by default, single precision
#device= whether the tensor lives on cpu or gpu
#requires_grad= tracks gradient
```

```python
rand_tensor
```

**Tensors need to have compatible data type, shape and device to be operable**
1. Device type refers to the location of the tensor, whether it lives on cpu or gpu (device="cuda")

```python
tensor_DJ= torch.arange(start=1, end=9, step=3, dtype=torch.float32)
tensor_DJ, tensor_DJ.dtype
```

```python
tensor_DJ+10
```

```python
tensor_DJ*10
```

```python
tensor_DJ-10
```

```python
torch.mul(tensor_DJ,10) #pytorch operator for element-wise matrix multiplication
```

```python
#matrix multiplication
#element-wise
tensor_DJNew=torch.rand_like(tensor_DJ,dtype=torch.float32)
print(tensor_DJNew)
elemMul = tensor_DJ*tensor_DJNew
elemMul
```

```python
matrixmul=torch.matmul(tensor_DJNew, tensor_DJ)
matrixmul
```
